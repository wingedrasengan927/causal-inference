**Q**: What's the difference between Association and Causality?

**A**: Association is when two quantities or random variables move together, whereas causality is when change in one variable causes change in another.

---

**Q**: What is the fundamental problem of causal inference?

**A**: The fundamental problem of causal inference is that you can never observe the same unit with and without treatment.

---

**Q**: In the context of MIMIR, how does a simple model trained only on control data mistake association and causation?

**A**: The control logic shows higher prices to users buying expensive flight tickets. So a user who has more purchase power buys expensive flight tickets and hence is shown higher prices. Since the user has high purchase power, he might even buy the higher-priced insurance. So the model might associate high plans with high attach rate, which is clearly not the case

---

**Q**: In the context of MIMIR, what is the treatment and the effect?

**A**: `ins_shown_per_pax` is the treatment, and `is_ins_purchased` is the effect

---

**Q**: In the context of MIMIR, give an example of a causal model for control logic

**A**: 
$$\text{price} \leftarrow f_t(u_t),$$

where $u_t$ is `fare_per_pax`, and

$$\text{isPurchased} \leftarrow f_y(T, u_y),$$

where $u_y$ is a set of unmodeled variables

**Q**: In the context of MIMIR, what's the difference between
$$\mathrm{E}(\text{isPurchased} \mid \text{price} = 149),$$
and
$$\mathrm{E}\left( \text{isPurchased} \mid \operatorname{do} \left( \text{price} = 149 \right) \right)$$

**A**: The average purchase rate among users who were actually shown the ₹149 plan vs. the average purchase rate if we forced the ₹149 plan on the entire population of users.

**Q**: How do you read $Y_{ti}$?

**A**: The potential outcome for user $i$ with treatment $t$

---

**Q**: In the context of MIMIR, given an example of individual treatment effect

**A**:
$$
\begin{align*}
\tau_i 
&= \text{isPurchased}_i \mid \operatorname{do} \left( \text{price} = 149 \right)  -  \text{isPurchased}_i \mid \operatorname{do} \left( \text{price} = 199 \right) \\[4pt]
&= Y_{149, i} - Y_{199, i}
\end{align*}
$$

---

**Q**: What is the fundamental problem of causal inference in terms of individual treatment effect?

**A**: You can never know the individual treatment effect because you only observe one of the potential outcomes

---

**Q**: How is the average treatment effect for a binary treatment defined?

**A**: 
$$
\begin{align*}
\text{ATE} &= \mathrm{E} \left[ \tau_i \right] \\[4pt]
&= \mathrm{E} \left[ Y_{1, i} - Y_{0, i} \right]
\end{align*}
$$

---

**Q**: How is the average treatment effect on the treated for a binary treatment defined?

**A**: 
$$
\begin{align*}
\text{ATT} &= \mathrm{E} \left[ \tau_i \mid T=1 \right] \\[4pt]
&= \mathrm{E} \left[ Y_{1, i} - Y_{0, i} \mid T=1 \right]
\end{align*}
$$

---

**Q**: How is the conditional average treatment effect for a binary treatment defined?

**A**: 
$$
\begin{align*}
\text{CATE} &= \mathrm{E} \left[ \tau_i \mid X=x \right] \\[4pt]
&= \mathrm{E} \left[ Y_{1, i} - Y_{0, i} \mid X=x \right],
\end{align*}
$$

where the random variable $X$ usually represents user features

---

**Q**: How are $\mathrm{E}[Y \mid T=t]$ and $\mathrm{E}[Y_t]$ related?

**A**: $\mathrm{E}[Y \mid T=t]$ is usually a biased estimator of $\mathrm{E}[Y_t]$

---

**Q**: Explain Bias in the context of MIMIR by giving an example

**A**: The average purchase rate of users who were shown the 199 plan through the control logic would not be equal to the true purchase rate of the 199 plan

---

**Q**: In the context of MIMIR, let ₹499 plan be the treatment ($T=1$) and ₹99 be the control ($T=0$). Explain the difference between $\mathrm{E}[Y_{499}]$ and $\mathrm{E}[Y_{499} \mid T=1]$?

**A**: $\mathrm{E}[Y_{499}]$ is the expected purchase rate of the ₹499 plan, while $\mathrm{E}[Y_{499} \mid T=1]$ is the expected purchase rate of the ₹499 plan among the users who were shown the ₹499 plan.

---

**Q**: In the context of MIMIR, let ₹499 plan be the treatment ($T=1$) and ₹99 be the control ($T=0$). Explain why
$$
Y_{499, i} \mid \left( T=1 \right) = Y_{i} \mid \left( T=1 \right)
$$

**A**: For the user who has been shown the ₹499 plan, the potential outcome of purchasing the plan is equal to the observed outcome

---

**Q**: In the context of MIMIR, let ₹499 plan be the treatment ($T=1$) and ₹99 be the control ($T=0$). How is association and causation between the treatment and the oucome measured

**A**: Association is measured as:
$$
\mathrm{E} \left[ Y \mid T=1 \right] -  \mathrm{E} \left[ Y \mid T=0 \right],
$$
while causation is measured as:
$$
\mathrm{E} \left[ Y \mid \operatorname{do}(t=1) \right] -  \mathrm{E} \left[ Y \mid \operatorname{do}(t=0) \right],
$$

---

**Q**: Intuitively, what does Bias represent in the context of Causal inference?

**A**: The Bias represents how the treated group and the control group differ, regardless of the treatment they've received

---

**Q**: In the context of MIMIR, let ₹499 plan be the treatment ($T=1$) and ₹99 be the control ($T=0$). Derive the bias equation

**A**: The association between the two groups is calculated as:
$$
\begin{align*}
\text{Association}
&= \mathrm{E} \left[ Y \mid T  = 1 \right] - \mathrm{E} \left[ Y \mid T  = 0 \right] \\[4pt]
&= \mathrm{E} \left[ Y_{499} \mid T  = 1 \right] - \mathrm{E} \left[ Y_{99} \mid T  = 0 \right] \\[4pt]
&= \mathrm{E} \left[ Y_{499} \mid T  = 1 \right] - \mathrm{E} \left[ Y_{99} \mid T  = 0 \right] + \\[4pt]
&\qquad \mathrm{E} \left[ Y_{99} \mid T  = 1 \right] - \mathrm{E} \left[ Y_{99} \mid T  = 1 \right] \\[4pt]
&= \mathrm{E} \left[ Y_{499} \mid T  = 1 \right] - \mathrm{E} \left[ Y_{99} \mid T  = 1 \right] + \\[4pt]
&\qquad \mathrm{E} \left[ Y_{99} \mid T  = 1 \right] - \mathrm{E} \left[ Y_{99} \mid T  = 0 \right] \\[4pt]
&= \text{ATT} + \text{Bias},
\end{align*}
$$

where
$$
\text{Bias} = \mathrm{E} \left[ Y_{99} \mid T  = 1 \right] - \mathrm{E} \left[ Y_{99} \mid T  = 0 \right]
$$

---

**Q**: Intuitively, what does the independence assumption mean?

**A**: It means the treated and untreated groups are comparable and indistinguishable

---

**Q**: In the context of MIMIR, let ₹499 plan be the treatment ($T=1$) and ₹99 be the control ($T=0$). Intuitively, when does the Bias term vanish?

**A**: When the group of users who are being shown the ₹499 plan is comparable to the group of users who are being shown the ₹99 plan

---

**Q**: In the context of MIMIR, let ₹499 plan be the treatment ($T=1$) and ₹99 be the control ($T=0$). What does the independence assumption mathematically mean?

**A**: 
$$
\mathrm{E} \left[ Y_{99} \mid T=1  \right] = \mathrm{E} \left[ Y_{99} \mid T=0  \right]
$$

---

**Q**: In the context of MIMIR, how can the independence assumption be realized?

**A**: By showing arms to users at random. This is already being done in the experiment `mimir_pure_exploration`
