**Q**: What are the three steps useful to tackle a causal inference problem?

**A**: 
1. Describe the problem with a graph.
2. Articulate what we would like to know.
3. Find the way to calculate the causal effect.

---

**Q**: What is the difference between $\mathrm{P} \mid T = A$ and $\mathrm{P} \mid \operatorname{do} ( T = A )$

**A**: $\mathrm{P} \mid T = A$ represents the observed conditional probability given $T = A$, while $\mathrm{P} \mid \operatorname{do} ( T = A )$ represents the hypothetical probability if everyone was treated with treatment A

---

**Q**: In a causal graph, when we intervene on a node $T$ by setting it to a value $A$, why do we remove the incoming arrows pointing to $T$?

**A**: Because the value of $T$ is fixed and it no longer depends on any other variables in the system.

---

**Q**: Given three discrete random variables: the treatment $T$, the outcome $Y$, and the confounder $S$, what is the adjustment formula for computing:

$$\mathrm{P}(Y = 1 \mid \operatorname{do} \left( T = A \right) )$$

**A**: $$\mathrm{P}(Y = 1 \mid \operatorname{do} \left( T = A \right)) = \sum_s \mathrm{P}(Y = 1 \mid T = A, S=s) \cdot \mathrm{P}(S = s)$$

---

**Q**: Given three discrete random variables: the treatment $T$, the outcome $Y$, and the confounder $S$, derive the adjustment formula for computing:

$$\mathrm{E}(Y  \mid \operatorname{do} ( T = A))$$

**A**:
By definition, we have:

$$
\begin{align*}
\mathrm{E}(Y \mid \operatorname{do} \left( T = A \right) )
&= \sum_y y \cdot \mathrm{P}(Y = y \mid \operatorname{do} (T = A)) \\[4pt]
&= \sum_y y \cdot \sum_s \mathrm{P}(Y = y \mid T = A, S=s) \mathrm{P}(S = s) \\[4pt]
&= \sum_s \sum_y y \cdot \mathrm{P}(Y = y \mid T = A, S=s) \mathrm{P}(S = s) \\[4pt]
&= \sum_s \mathrm{P}(S = s) \sum_y y \cdot \mathrm{P}(Y = y \mid T = A, S=s) \\[4pt]
&= \sum_s \mathrm{E} \left[ Y \mid T = A, S=s \right] \cdot \mathrm{P}(S = s)
\end{align*}$$

**Q**: How is the ATE of two treatments $A$ and $B$ on a binary outcome $Y$ defined

**A**: It's defined as the difference:
$$
\mathrm{P} (Y = 1 \mid \operatorname{do}(T=A)) - \mathrm{P} (Y = 1 \mid \operatorname{do}(T=B))
$$

---

**Q**: How is the ATE of two treatments $A$ and $B$ on a continuous outcome $Y$ defined

**A**: It's defined as the difference:
$$
\mathrm{E} (Y \mid \operatorname{do}(T=A)) - \mathrm{E} (Y \mid \operatorname{do}(T=B))
$$

---

**Q**: Mathematically, what does the positivity assumption state

**A**: We say that the positivity assumption holds if for all the values confounders may take, $s = (s_1, \ldots, s_p)$, and for each treatment, $T = t$, we have that

$$0 < \mathrm{P}_0 (T = t \mid S = s) < 1$$

Here, $\mathrm{P}_0$ represents the true data-generation distribution, not the empirical distribution obtained from the observed data.

---

**Q**:  In what type of experimental design does $$\mathrm{P}(Y=1 \mid \operatorname{do}(T=A))$$ equal $$\mathrm{P}(Y=1 \mid T=A)$$

**A**: In a Randomized Controlled Trial (RCT) or A/B test

**Q**: What is a predictive outcome variable?

**A**: A predictive outcome variable predicts the outcome but doesn't influence the treatment

---

**Q**: Given three discrete random variables: the treatment $T$, the outcome $Y$, and a predictive outcome variable $P$, what is the adjustment formula for computing:

$$\mathrm{P}(Y = 1 \mid \operatorname{do} \left( T = A \right) )$$

**A**: Intuitively, since $P$ isn't a confounder, we have:

$$\mathrm{P}(Y = 1 \mid \operatorname{do} \left( T = A \right) ) = \mathrm{P}(Y = 1 \mid  T = A )$$

**Q**: Given three discrete random variables: the treatment $T$, the outcome $Y$, and a predictive outcome variable $P$, derive the adjustment formula for computing:

$$\mathrm{P}(Y = 1 \mid \operatorname{do} \left( T = A \right) )$$

**A**: Since $P$ influences the outcome $Y$, we have:

$$
\begin{align*}
\mathrm{P}(Y = 1 \mid \operatorname{do} \left( T = A \right) ) 
&= \sum_p \mathrm{P}(Y = 1 \mid T = A, P=p ) \cdot \mathrm{P}(P = p)
\end{align*}
$$

Since $T$ and $P$ are independent, the above expression can be simplified as:

$$
\begin{align*}
\frac{1}{\mathrm{P}(T=A)} \cdot \sum_p \mathrm{P}(Y = 1 \mid T = A, P=p ) \cdot \mathrm{P}(T=A, P = p)
\end{align*}
$$

By definition of conditional probability, we have:
$$
\mathrm{P}(Y = 1 \mid T = A, P=p ) = \frac{\mathrm{P}(Y = 1, T = A, P=p )}{\mathrm{P}(T = A, P=p )}
$$

Now the expression can be simplified as:
$$
\begin{align*}
\frac{1}{\mathrm{P}(T=A)} \cdot \sum_p \mathrm{P}(Y = 1, T = A, P=p )
\end{align*}
$$

By using the definition of marginal probability and conditional probability, we get:
$$
\begin{align*}
\frac{1}{\mathrm{P}(T=A)} \cdot \sum_p \mathrm{P}(Y = 1, T = A, P=p )
&= \frac{1}{\mathrm{P}(T=A)} \cdot \mathrm{P}(Y = 1, T = A) \\[4pt]
&= \mathrm{P}(Y = 1 \mid T = A)
\end{align*}
$$

**Q**: Given three discrete random variables: the treatment $T$, the outcome $Y$, and a confounder $Z$, what is the adjustment formula for computing:

$$\mathrm{P}(Y = 1 \mid \operatorname{do} \left( T = A \right), Z=z )$$

**A**: We have

$$\mathrm{P}(Y = 1 \mid \operatorname{do} \left( T = A \right), Z=z ) = \mathrm{P}(Y = 1 \mid T = A, Z=z )$$
