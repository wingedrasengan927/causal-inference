**Q**: What is experimental data

**A**: Data obtained through an A/B experiment is called experimental data

---

**Q**: What is observational data

**A**: Data obtained without any experimental setup is called observational data

---

**Q**: In an A/B experiment, what can you say about the A and B groups

**A**: Group A and group B will have the same distribution of population characteristics.

---

**Q**: What is the key characteristic of an A/B test?

**A**: Treatment assignment is done at random

---

**Q**: In words, how is the average treatment effect in an A/B test defined

**A**: It's defined as the difference in the average results between the two groups

---

**Q**: In words, what is a confounder variable

**A**: It's a variable that influences both the treatment assignment and the outcome

---
