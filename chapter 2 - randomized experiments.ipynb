**Q**: In the context of MIMIR, let ₹499 plan be the treatment ($T=1$) and ₹99 be the control ($T=0$). What does $Y_{99} \perp T$ indicate?

**A**: It indicates that knowing which plan a user was actually offered gives us no information about how they would have responded to the ₹99 plan

---

**Q**: In practice, how would you make sure the treatment and the control group are roughly comparable?

**A**: By randomizing the treatment assignment

---

**Q**: In practice, how would you check if the randomized assignment of treatment is properly done?

**A**: Compute the normalized difference for pretreatment variables and check if its absolute value isn't too high
$$
\frac{\hat{\mu}_{tr} - \hat{\mu}_{co}}{\sqrt{\left( \hat{\sigma}_{tr}^2 - \hat{\sigma}_{co}^2 \right) / 2}}
$$
