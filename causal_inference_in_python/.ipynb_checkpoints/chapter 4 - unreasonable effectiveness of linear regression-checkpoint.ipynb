**Q**: In a linear regression setting, what is the interpretation of $\beta_1$ - the coefficient of the treatment?

**A**: It can be viewed as the change in outcome, given a small change in the treatment, while holding fixed all other variables in the model

---

**Q**: How do you estimate ATE when the treatment is continuous and the treatment assignment is random?

**A**: $$\text{ATE} = \frac{\partial E[Y \mid T=t]}{\partial t}$$

---


**Q**: Explain the Debiasing step in the FWL Theorem


**A**: First we find the least squares approximate solution $\hat{\beta}$ to the OLS problem:

$$
\text{minimize} \quad \| X \beta - T \|^2,
$$

where $X$ is the matrix of confounders (with the first column being a constant vector of ones) and $T$ is the treatment variable. Finally, the residuals can be computed as:

$$
\tilde{T} = T - X\hat{\beta}
$$

---

**Q**: Derive the formula for the covariance between a confounder variable $X$ and the residual $\tilde{T}$.

**A**: The sample covariance is calculated as:

$$\operatorname{Cov} \left( X, \tilde{T} \right) = \frac{1}{N} \sum_{i=1}^N \left( X_{i} - \bar{X} \right) \left( \tilde{T}_i - \bar{\tilde{T}} \right)$$

By the OLS property, we have $\bar{\tilde{T}} = 0$. Substituting this into the equation, we get:

$$\begin{align*}
\operatorname{Cov} \left( X, \tilde{T} \right) 
&= \frac{1}{N} \left( \sum_{i=1}^N  X_{i} \tilde{T}_i - \bar{X} \sum_{i=1}^N \tilde{T}_i \right) \\[4pt]
&= \frac{1}{N} \sum_{i=1}^N  X_{i} \tilde{T}_i
\end{align*}$$

---

**Q**: Let $\tilde{T}$ be the residuals obtained from the FWL theorem. Explain why the correlation between each confounder variable and $\tilde{T}$ is zero.

**A**: By property of OLS, each confounder vector $X_j$ (a column of $X$) is orthogonal to the residual vector $\tilde{T}$. Therefore, their dot product is zero. Substituting this into the covariance formula, we get:

$$\operatorname{Cov} \left( X_j, \tilde{T} \right) = \frac{1}{N} \sum_{i=1}^N  X_{ij} \tilde{T}_i = 0 $$

Thus, the correlation is zero.

---

**Q**: What is the purpose of the denoising step in the FWL theorem

**A**: To reduce the variance of the treatment effect estimate

---

**Q**: Explain the final step in the FWL theorem

**A**: In the final step, we just regress $\tilde{Y}$ on $\tilde{T}$. The parameter estimate for $\tilde{T}$ gives the estimate of the causal effect of the treatment on the outcome


---

**Q**: What's the first check you need to do before applying the FWL theorem

**A**: Check if the causal relation between the treatment and the outcome is linear.

---

 **Q**: Suppose you regress the outcome $Y$ on the treatment $T$ and the
 confounders $X$. Then, according to the FWL theorem, what does $\tau$ -
 the coefficient of $T$ represent? 

 **A**: The causal treatment effect estimate
