**Q**: How is propensity score defined?

**A**: The propensity score is simply the conditional probability of receiving a specific treatment, given a set of covariates $X$

---

**Q**: Why does it make sense to compare two users with the same propensity score?

**A**: If two users have the exact same propensity score, the only reason they received different treatments is by pure chance. Thus, conditioned on this score, the treatment assignment is as good as random

---

**Q**: What is the intuitive idea behind propensity score matching?

**A**: Find pairs of units from different treatment groups that have similar propensity scores. Then, use the actual outcome of one unit to impute the unobserved potential outcome for its match.

---

**Q**: Give an example of propensity score matching in the context of MIMIR

**A**: Suppose two users have purchased a flight with a fare of ~₹2500. Both have an equal probability of receiving the ₹129 treatment. However, due to the random nature of treatment assignment, it may be the case that one user receives a treatment of ₹199. If we know the outcome of this user, we can use it to impute the potential outcome of the other user had they been shown the ₹199 plan.

---

**Q**: What is the intuitive idea behind inverse propensity weighing?

**A**: The idea is to reweight the data by giving high importance to treated units that look like the untreated, and to untreated units that look like the treated.

---

**Q**: In inverse propensity weighting, a treated unit that looks like an untreated unit is a good estimate of what?

**A**: $Y_1 \mid T=0$

---

**Q**: How can you estimate the expected potential outcome of a specific treatment using the IPW estimator?

**A**: 
$$
\mathrm{E} \left[ Y_t \right] = \mathrm{E} \left[ \frac{\mathbf{I}(T=t) Y}{\mathrm{P} \left( T=t \mid X \right)} \right]
$$

---

**Q**: How can you estimate the ATE using the IPW estimator?

**A**: 

$$
\begin{align*}
ATE &= \mathrm{E} \left[ Y_1 \right] - \mathrm{E} \left[ Y_0 \right] \\[6pt]
&= \mathrm{E} \left[ \frac{\mathbf{I}(T=1) Y}{\mathrm{P} \left( T=1 \mid X \right)} \right] - \mathrm{E} \left[ \frac{\mathbf{I}(T=0) Y}{\mathrm{P} \left( T=0 \mid X \right)} \right]
\end{align*}
$$

---

**Q**: What is the formula for stabilized IPW

**A**: 

$$
w = \frac{\mathrm{P}(T = t)}{\mathrm{P}(T = t \mid X)}
$$

---

**Q**: When should we use stabilized propensity weights?

**A**: When standard propensity weights are extremely large

---

**Q**: In general, when does the IPW estimator have high variance?

**A**: When the propensity scores for some units are too close to zero or one, which creates extremely large weights.

---

**Q**: Suppose a model estimates propensity scores too accurately. In this case, the IPW estimator has high ___ and low ___?
 
**A**: High variance and low bias.

---

**Q**: Suppose a model estimates propensity scores too accurately. In this case, why does the IPW estimator have high variance?
 
**A**: The propensity scores for some units would be close to zero or one, which results in an explosion of inverse propensity weights.

---

**Q**: What is the disadvantage of a model that estimates propensity scores too accurately?
 
**A**: We miss out on the rare but important units where the propensity score and the actual treatment assigned disagree.

---

**Q**: When is the positivity assumption violated in the context of IPW?

**A**: When there are no treated samples with low propensity scores, and no untreated samples with high propensity scores

---

**Q**: What is the intuitive idea behind Doubly Robust estimator?

**A**: DR estimator combines propensity score and linear regression in a way that only one of them needs to be rightly specified.

---

**Q**: In practice, how would you fit a potential outcome model $\hat{m}(X)$ for estimating $\mathrm{E} \left[ Y_t \right]$?

**A**: Fit a linear regression model only on the samples which received treatment $t$, and then make predictions on the entire dataset

---

**Q**: What is the mathematical expression for the Doubly Robust estimator

**A**: 

$$
\hat{\mu}_t^\text{DR} \left( \hat{m}, \hat{e} \right) = 
\frac{1}{N} \sum \hat{m} \left( X \right) + \frac{1}{N} \sum \left[ \frac{\mathbf{I}(T = t)}{\hat{e}(x)} \cdot \left( Y - \hat{m}(X) \right) \right],
$$

where $\hat{m}(X)$ is a model for $\mathrm{E} \left[ Y_t \mid X \right]$ and $\hat{e}(X)$ is a propensity score model for $\mathrm{P}(T \mid X)$

---

**Q**: In a doubly robust estimator, what happens when the propensity score model $\hat{e}(X)$ is wrong but the outcome model $\hat{m}(X)$ is correct?

**A**: The second term converges to zero, and you would be left with the first term, which is just the correct outcome model

---

**Q**: In a doubly robust estimator, what happens when the propensity score model $\hat{e}(X)$ is correct but the outcome model $\hat{m}(X)$ is wrong?

**A**: The DR estimator acts just like the IPW estimator
