**Q**: What is a visual trick to check if a causal graph has bias?

**A**: Create a causal graph that is just like the original one, but with the causal relationship between treatment and outcome removed. If they are still connected in this graph, it must be due to a noncausal path, which indicates the presence of bias

**Q**: State the conditional independence assumption

**A**: The conditional independence assumption states that, if you compare units with the same level of covariates, their potential outcomes will be, on average, the same.

---

**Q**: How does the deterministic control logic in MIMIR violate the positivity assumption?

**A**: We cannot calculate the ATE for a specific `fare_per_pax` group (e.g., ₹4999-₹5999) because the policy assigns only one price plan per group, meaning alternative treatments are never observed

---

**Q**: How would you calculate the unbiased estimate of the ATE in the presence of bias?

**A**: Segment the data into groups with the same covariates. Calculate the difference between treated and control within each group, then take the weighted average of these differences using the group sizes as weights.

**Q**: When does confounding bias occur?

**A**: Confounding bias usually occurs when the treatment and the outcome share a common cause.
